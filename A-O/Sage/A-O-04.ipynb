## Решётки — определение и примеры

- $\mathbb{R}^n$ — евклидово векторное пр-во; $\mathbf{v}_1,\dots,\mathbf{v}_k\in\mathbb{R}^n$ — лин. независимы;
- (Евклидова) решётка $\leftrightharpoons$ $L = \left\{\sum_{i=1}^n a_i \mathbf{v}_i : a_i \in \mathbb{Z}\right\} \subset \mathbb{R}^n$;
- $k$ — ранг $L$; если $k=n$, то $L$ — полная в $\mathbb{R}^n$.

```python
# построение точек решётки по базисной матрице B: xB, x in {-N..N}^n, 
import itertools

def lattice_points(B, N):
    n = B.nrows()
    return [vector(a)*B for a in itertools.product(range(-N, N+1), repeat=n)]

def plot_lattice(B, N, **kwds):
    return point2d(lattice_points(B, N), size=10, **kwds)

B_hex = matrix(RDF, [[1, 0], [1/2, sqrt(3)/2]])
plot_lattice(B_hex, 5, color='darkgreen', axes=False, aspect_ratio=1)
```

```python
# у одной решётки много базисов: (1,0),(0,1) и (1,0),(1,1) порождают одну и ту же Z^2.
# можно сравнить множество точек (в шаре конечного радиуса)
def in_ball(pts, R):
    return [p for p in pts if p.norm() <= R]

B1 = matrix(ZZ, [[1, 0], [0, 1]])
B2 = matrix(ZZ, [[1, 0], [1, 1]])
N = 10
R = 5

pts1 = set(tuple(v) for v in in_ball(lattice_points(B1, 20), R))
pts2 = set(tuple(v) for v in in_ball(lattice_points(B2, 20), R))
print("одинаковые множества точек в шаре радиуса", R, ":", pts1 == pts2, "| точек:", len(pts1))
```

```python
plot_lattice(B1, N, color='blue', axes=False, aspect_ratio=1) + \
circle((0, 0), R, color='red', linestyle='dashed', thickness=2)
```

```python
plot_lattice(B2, N, color='blue', axes=False, aspect_ratio=1) + \
circle((0, 0), R, color='red', linestyle='dashed', thickness=2)
```

```python
# контрпример: 1 и sqrt(2) линейно зависимы над Q, но не порождают решётку в R^1 --
# множество {a + b*sqrt(2) : a,b in Z} всюду плотно; при росте N минимальный зазор -> 0
def points_1_sqrt2(N):
    s = sqrt(2).n(200)
    return sorted(a + b*s for a in range(-N, N+1) for b in range(-N, N+1))

for N in [2, 5, 10, 20]:
    pts = points_1_sqrt2(N)

pts5 = points_1_sqrt2(5)
list_plot([(p, 0) for p in pts5], size=10, axes=False, aspect_ratio='automatic')
```

```python
# Решётки в R^3
def plot_lattice3d(B, N, point_color='blue', basis_color='red'):
    pic = point3d(lattice_points(B, N), size=15, color=point_color)
    for v in B.rows():
        pic += arrow3d((0, 0, 0), tuple(v), color=basis_color, width=1)
    pic.show(viewer='threejs', online=True)

# тривиальная
B_cubic = matrix(ZZ, [[1, 0, 0], [0, 1, 0], [0, 0, 1]])
plot_lattice3d(B_cubic, 2)
```

```python
# "гранецентрированная" (FCC, face-centered cubic)
B_fcc = matrix(ZZ, [[1, 1, 0], [1, 0, 1], [0, 1, 1]])
plot_lattice3d(B_fcc, 2, point_color='darkgreen')
```

```python
# "объёмноцентрированная" (BCC, body-centered cubic) -- "двойственная" к FCC
B_bcc = matrix(QQ, [[1, 0, 0], [0, 1, 0], [1/2, 1/2, 1/2]])
plot_lattice3d(B_bcc, 2, point_color='purple')
```

## *Модуль над кольцом

- Векторное пр-во $V$ над полем $K$ $\leftrightharpoons$ абелева группа $(V,+)$ с умножением на скаляры из $K$: $a(\mathbf{u}+\mathbf{v})=a\mathbf{u}+a\mathbf{v}$, $(a+b)\mathbf{v}=a\mathbf{v}+b\mathbf{v}$, $(ab)\mathbf{v}=a(b\mathbf{v})$, $1\cdot\mathbf{v}=\mathbf{v}$;
- Модуль $M$ над (коммутативным) кольцом $R$ ($R$-модуль) $\leftrightharpoons$ те же аксиомы, но скаляры — из кольца $R$;
- Решётка $L\subset\mathbb{R}^n$ — $\mathbb{Z}$-модуль;
- Любую коммутативную (абелеву) группу $G$ можно рассматривать как $\mathbb{Z}$-модуль: умножение на $a\in\mathbb{Z}$: $a\cdot g = g+g+\dots+g$;
- Если $G$ — конечная, $|G|=n$, то $\forall\ g\in G\ n\cdot g=0$;
- В общем случае: может найтись $r\in R\setminus\{0\}$ и $\mathbf v\in M\setminus\{0\}$ такие, что $r\mathbf v=0$ (элементы кручения) $\Rightarrow$ понятие лин. независимости теряет смысл;
- Модуль, обладающий базисом (т.е. через который выражаются все элементы), называется свободным, число элементов базиса — рангом;
- Т.о. более точно: решётка $L\subset\mathbb{R}^n$ — свободный $\mathbb{Z}$-модуль ранга $k$.

```python
# пример: решётка как свободный Z-модуль ранга k. span(ZZ, B) и (ZZ^3).submodule(...) --
# два способа задать тот же подмодуль; FCC как подрешётка Z^3 индекса 2 (сравните с det(B_fcc))
M_span = span(ZZ, B_fcc.rows())
M_sub = (ZZ^3).submodule(B_fcc.rows())
print("span(ZZ, B_fcc) == (ZZ^3).submodule(...):", M_span == M_sub)

print("det(B_fcc) =", B_fcc.det())
print("[Z^3 : FCC] =", (ZZ^3 / M_sub).cardinality())
```

```python
# пример: Z/2Z как Z-модуль с кручением -- 2*x = 0 при x != 0, у модуля с кручением
# базиса нет (x нельзя продолжить до базиса свободного модуля)
Mtor = Zmod(2)
for x in Mtor:
    print(f"2*{x} = {2*x}")
```

```python
# пример: произвольная конечная абелева группа G, |G|=n 
# любой элемент g обнуляется умножением на n (n*g=0)
G = AdditiveAbelianGroup([4, 6])
n = G.order()
print("|G| =", n)
for g in [G.random_element() for _ in range(5)]:
    print(f"{n}*{g.vector()} = {(n*g).vector()}")
```

```python
# пример: Q как Z-модуль -- без кручения, но не свободен
# любые два элемента x,y in Q лин. зависимы над Z: 
# найдутся целые a,b, не оба нуля, a*x+b*y=0
x, y = QQ(7/3), QQ(-11/5)
ratio = x / y
a, b = ratio.denominator(), -ratio.numerator()
print(f"a={a}, b={b}:  a*x + b*y = {a*x + b*y}")

# без кручения: n*q=0 c n != 0 влечёт q=0 (проверка в интервале)
print("Q без кручения:", all(n * QQ((p, q)) != 0 for n in range(1, 6)
                              for p in range(1, 4) for q in range(1, 4)))
```

## Фундаментальный параллелепипед и объём

- Фундаментальный параллелепипед решётки $\leftrightharpoons$ $P(L) = \left\{\sum_{i=1}^n \alpha_i \mathbf{v}_i : \alpha_i \in [0,1)\right\}$.
- Объём решётки $\leftrightharpoons$ $\mathrm{vol}(L)=\mathrm{vol}(P(L))$.
- Лемма: $\mathrm{vol}(L)=|\det B|$, где $B$ — матрица из $\mathbf{v}_1,\dots,\mathbf{v}_n$; в частности, $\mathrm{vol}(L)$ не зависит от выбора базиса $L$.

```python
# vol(L) = |det B| для базисной матрицы B
B = matrix(QQ, [[3/2, 0], [1/2, 5/4]])
print("vol(L) = |det B| =", abs(B.det()))
```

```python
# смена базиса решётки <-> умножение слева на унимодулярную матрицу (det U = +-1);
# объём (и множество точек решётки) при этом не меняется
U = matrix(ZZ, [[2, 1], [1, 1]])
Bp = U * B
print("det U =", U.det())
print("det B =", B.det(), " det(U*B) =", Bp.det())

R = 6
pts_B = set(tuple(v.n(10)) for v in in_ball(lattice_points(B, 40), R))
pts_Bp = set(tuple(v.n(10)) for v in in_ball(lattice_points(Bp, 40), R))
print("B и U*B порождают одно и то же множество точек в шаре:", pts_B == pts_Bp)
```

```python
# P(L) для двух разных базисов одной решётки: разная форма параллелепипеда,
# один и тот же объём (площадь) и одна и та же решётка точек
def fundamental_parallelepiped(B, **kwds):
    v1, v2 = B.rows()
    return polygon2d([(0, 0), tuple(v1), tuple(v1 + v2), tuple(v2)], fill=True, alpha=0.35, **kwds)

B_other = U * B  # другой базис той же решётки, что и B

pic1 = plot_lattice(B, 4, color='black', axes=False, aspect_ratio=1) + \
       fundamental_parallelepiped(B, color='blue')
pic2 = plot_lattice(B, 4, color='black', axes=False, aspect_ratio=1) + \
       fundamental_parallelepiped(B_other, color='red')

print("площадь P(B)      =", abs(B.det()).n())
print("площадь P(U*B)    =", abs(B_other.det()).n())
pic1 + pic2
```

```python
# индекс подрешётки = отношение объёмов: [L:M] = vol(M)/vol(L);
# пример -- FCC как подрешётка Z^3 индекса 2, объём FCC = 2
B_cubic = matrix(ZZ, [[1, 0, 0], [0, 1, 0], [0, 0, 1]])
B_fcc = matrix(ZZ, [[1, 1, 0], [1, 0, 1], [0, 1, 1]])
M_fcc = (ZZ^3).submodule(B_fcc.rows())

print("vol(Z^3) =", abs(B_cubic.det()), " vol(FCC) =", abs(B_fcc.det()))
print("[Z^3 : FCC] =", (ZZ^3 / M_fcc).cardinality(),
      " == vol(FCC)/vol(Z^3) =", abs(B_fcc.det()) / abs(B_cubic.det()))
```

```python
# ещё один, многомерный пример связи индекса подрешётки и объёма: M subset Z^4
B4 = matrix(ZZ, [[2, 0, 0, 0], [0, 3, 0, 0], [0, 1, 1, 0], [1, 0, 0, 1]])
M4 = (ZZ^4).submodule(B4.rows())

print("vol(M4) = |det B4| =", abs(B4.det()))
print("[Z^4 : M4] =", (ZZ^4 / M4).cardinality())
```

## Некоторые характеристики решёток

- Пусть $\mathbf{u}\cdot \mathbf{v} = (\mathbf{u},\mathbf{v}) = \sum_i u_iv_i$ — «обычное» скалярное произведение;
- $L$ — целая $\leftrightharpoons$ $\forall\ \mathbf{u}, \mathbf{v} \in L\ \mathbf{u}\cdot \mathbf{v}\in\mathbb{Z}$;
- $L$ — унимодулярная $\leftrightharpoons$ целая и $\mathrm{vol}(L)=1$;
- $L$ — чётная $\leftrightharpoons$ целая и $|\mathbf{v}|^2=\mathbf{v}\cdot \mathbf{v}$ чётно для всех $\mathbf{v} \in L$.

```python
# проверять целость/чётность достаточно на базисе (удобнее по "матрице Грама" G = B*B^T):
# L целая <=> все v_i . v_j in Z; чётная <=> вдобавок все |v_i|^2 чётны
def gram(B):
    return B * B.transpose()

def is_integral(B):
    return all(g in ZZ for g in gram(B).list())

def is_unimodular(B):
    return is_integral(B) and abs(B.det()) == 1

def is_even(B):
    if not is_integral(B):
        return False
    G = gram(B)
    return all(ZZ(G[i, i]) % 2 == 0 for i in range(G.nrows()))

def report(name, B):
    print(f"{name:22s} целая={is_integral(B)!s:5s} унимодулярная={is_unimodular(B)!s:5s} чётная={is_even(B)}")
```

```python
# целые решётки: Z^2 и Z^n -- целые; гексагональная -- не целая
B_Z2 = identity_matrix(ZZ, 2)
B_Zn = identity_matrix(ZZ, 4)
B_hex = matrix(AA, [[1, 0], [1/2, sqrt(3)/2]])

report("Z^2", B_Z2)
report("Z^4", B_Zn)
report("гексагональная", B_hex)
```

```python
# масштабирование гексагональной на sqrt(2) "чинит" целость:
# v1=(sqrt2,0), v2=(1/sqrt2, sqrt3/sqrt2) -- целая (и даже чётная, см. ниже)
B_hex_sqrt2 = matrix(AA, sqrt(2) * B_hex)
report("sqrt2*гексагональная", B_hex_sqrt2)
print(gram(B_hex_sqrt2))
```

```python
# унимодулярность: Z^n -- унимодулярна
# базис v1=(1,0), v2=(0,2) задаёт целую, но не унимодулярную решётку
# чётность: Z^n не чётна (|(1,0,...,0)|^2=1 -- нечётно), sqrt2*гексагональная -- чётна (ещэ раз)
B_skew = matrix(ZZ, [[1, 0], [0, 2]])
report("Z^2 (ещё раз)", B_Z2)
report("<(1,0),(0,2)>", B_skew)
report("Z^4", B_Zn)
report("sqrt2*гексагональная", B_hex_sqrt2)
```

```python
# L = {x in Z^n: sum x_i чётно} -- чётная решётка (это D_n, см ниже, базис: e_i - e_{i+1}, e_{n-2}+e_{n-1})
n = 4
rows = [vector(ZZ, [1 if k == i else (-1 if k == i + 1 else 0) for k in range(n)]) for i in range(n - 1)]
rows.append(vector(ZZ, [0]*(n - 2) + [1, 1]))
B_Dn = matrix(ZZ, rows)

report(f"D_{n}", B_Dn)
print("[Z^n : D_n] =", (ZZ^n / (ZZ^n).submodule(B_Dn.rows())).cardinality())
```

```python
# для целой L: vol(L)^2 = det(v_i . v_j) in Z (на примере D_4)
G = gram(B_Dn)
print("det G =", G.det(), " in ZZ:", G.det() in ZZ)
print("vol(L) = |det B| =", abs(B_Dn.det()), " vol(L)^2 =", abs(B_Dn.det())^2)
```

## Приводимые и неприводимые решётки

- Пусть $L_1, L_2 \subset L$ — подрешётки $L$ (т.е. сами являются решётками);
- $L=L_1\perp L_2$ (ортогональная прямая сумма) $\leftrightharpoons$ каждый $\mathbf{v}\in L$ единственным образом записывается как $\mathbf{v}=\mathbf{v}_1+\mathbf{v}_2$, $\mathbf{v}_i\in L_i$, и $\mathbf{v}_1\cdot\mathbf{v}_2=0$ для всех $\mathbf{v}_1\in L_1$, $\mathbf{v}_2\in L_2$;
- $L$ — неприводимая $\leftrightharpoons$ $L\neq\{0\}$ и не раскладывается в $\perp$-сумму двух ненулевых решёток;
- Теорема (Эйхлер): любая решётка раскладывается в $\perp$-сумму неприводимых, причём единственным образом с точностью до порядка слагаемых.

```python
# Z (+) Z = Z^2: direct_sum совпадает с IntegralLattice(2) по матрице Грама
L1 = IntegralLattice(1)
L2 = IntegralLattice(1)
L_sum = L1.direct_sum(L2)
L_Z2 = IntegralLattice(2)

print("матрица Грама (Z (+) Z):\n", L_sum.gram_matrix())
print("матрица Грама (IntegralLattice(2)):\n", L_Z2.gram_matrix())
```

```python
# почему гексагональная неприводима: у Z^2 = aZ (+) bZ только 4 кратчайших вектора
# (+-a,0),(0,+-b), а у sqrt2*гексагональной -- 6 (как и у самой гексагональной);
# в ⊥-сумме двух решёток ранга 1 кратчайших векторов может быть только 2 или 4
L_Z2_short = L_Z2.short_vectors(2)   # список по нормам: [норма 0], [норма 1], ...
L_hex_sqrt2 = IntegralLattice(matrix([[2, 1], [1, 2]]))
L_hex_short = L_hex_sqrt2.short_vectors(3)

print("Z^2:                 кратчайших (нормы 1) =", len(L_Z2_short[1]))
print("sqrt2*гексагональная: кратчайших (нормы 2) =", len(L_hex_short[2]))
```

```python
# визуализация L = Z (+) (sqrt2*гексагональная) в R^3: Z вдоль оси x,
# sqrt2*гексагональная -- в ортогональной ей плоскости yz
s2, s3 = RDF(sqrt(2)), RDF(sqrt(3))
B_sum = matrix(RDF, [
    [1, 0, 0],
    [0, s2, 0],
    [0, s2/2, s2*s3/2],
])
plot_lattice3d(B_sum, 2, point_color='teal')
```

## «Одинаковые» решётки: изометрия и подобие

- Решётки $L\cong L'$ (изометричны) $\leftrightharpoons$ $L'=\varphi(L)$ для ортогонального $\varphi:\mathbb{R}^n\to\mathbb{R}^n$;
- Изометричность сохраняет основные характеристики: объем, целостность, чётность, унимодулярность, ... — это инварианты решётки.
- $L\sim L'$ (подобны) $\leftrightharpoons$ $L'\cong c\,L$ для некоторого $c>0$ (изометрия с точностью до масштаба, не сохраняет инварианты, но сохраняет структуру).
- На языке матриц: пусть $B$, $B'$ — матрицы базисов $L$, $L'$ (по строкам),
    - $L\cong L' \Leftrightarrow B'=UBO$, где $U\in GL_n(\mathbb{Z})$ (невырожденная), $O\in O(n)$ (ортогональная);
    - $L\sim L' \Leftrightarrow B'=c\,U\,B\,O$.

```python
# пример: Z^2 и L' с базисом (1/sqrt2,1/sqrt2),(-1/sqrt2,1/sqrt2) -- изометричны
# (поворот на 45 градусов): O ортогональна, и B*O даёт в точности базис L'
O = matrix(AA, [[1, 1], [-1, 1]]) / sqrt(AA(2))
print("O*O^T == I:", O * O.transpose() == identity_matrix(AA, 2))

B = identity_matrix(AA, 2)
Bp = B * O
target = matrix(AA, [[1/sqrt(AA(2)), 1/sqrt(AA(2))], [-1/sqrt(AA(2)), 1/sqrt(AA(2))]])
print("B*O совпадает с базисом L':", Bp == target)
```

```python
# наложение: Z^2 (синим) и L' = Z^2*O (красным) на одной плоскости
# это один и тот же набор точек, просто повёрнутый на 45 градусов вокруг начала координат
plot_lattice(matrix(RDF, B), 4, color='blue', axes=False, aspect_ratio=1) + \
plot_lattice(matrix(RDF, Bp), 4, color='red', axes=False, aspect_ratio=1)
```

```python
# пример: гексагональная и sqrt2*гексагональная -- подобны (c=sqrt2), но не изометричны:
# vol растёт в c^2 раз (инвариант нарушен), а vol/vmin^2 (инвариант подобия) сохраняется
def shortest_vectors(B, N=5, tol=1e-9):
    pts = lattice_points(matrix(RDF, B), N)
    nz = [p for p in pts if p.norm() > tol]
    m = min(p.norm() for p in nz)
    return m, [p for p in nz if abs(p.norm() - m) < tol]

c = sqrt(AA(2))
B_hex2 = c * B_hex
print("B_hex2 == c*B_hex:", B_hex2 == c * B_hex)
print("vol(hex) =", abs(B_hex.det()).n(), " vol(sqrt2*hex) =", abs(B_hex2.det()).n(),)
print("отношение =", (abs(B_hex2.det()) / abs(B_hex.det())).n(), " (c^2 =", (c^2).n(), ")")

# нормирование на кратчайший вектор
m_hex, sv_hex = shortest_vectors(B_hex)
m_hex2, sv_hex2 = shortest_vectors(B_hex2)
print("vol/vmin^2 (hex)      =", abs(B_hex.det()).n() / m_hex^2)
print("vol/vmin^2 (sqrt2*hex) =", abs(B_hex2.det()).n() / m_hex2^2)
```

```python
# наложение: гексагональная (зелёным) и sqrt2*гексагональная (оранжевым) на одной плоскости
# видно, что ориентация и углы совпадают (одна "форма"), просто вторая решётка более редкая
plot_lattice(matrix(RDF, B_hex), 4, color='darkgreen', axes=False, aspect_ratio=1) + \
plot_lattice(matrix(RDF, B_hex2), 3, color='orange', axes=False, aspect_ratio=1)
```

```python
# пример: Z^2 и гексагональная -- не изометричны и не подобны: посчитаем кратчайшие
# векторы перебором -- разное число (4 против 6) и разное vol/vmin^2 (инвариант подобия)
m_Z2, sv_Z2 = shortest_vectors(identity_matrix(QQ, 2))
m_hex, sv_hex = shortest_vectors(B_hex)

print("Z^2:        vmin =", m_Z2, " # кратчайших =", len(sv_Z2), " vol/vmin^2 =", 1.0 / m_Z2^2)
print("гексаг.:    vmin =", m_hex, " # кратчайших =", len(sv_hex),
      " vol/vmin^2 =", abs(B_hex.det()).n() / m_hex^2)
print("не изометричны (разное # кратчайших) и не подобны (разный vol/vmin^2)")
```

## Решётки корней

- $L$ — решётка корней $\leftrightharpoons$ $L$ — целая и порождена векторами длины $\sqrt2$;
- Т.е. если $R_L=\{\mathbf{r}\in L:\ |\mathbf{r}|^2=\mathbf{r}\cdot \mathbf{r}=2\}$, то $\forall\ \mathbf{v}\in L\ \mathbf{v}=\sum a_i \mathbf{r}_i,\ \mathbf{r}_i\in R_L, a_i\in\mathbb{Z}$;
- ($\mathbf{r}\in R_L$ называются корнями $L$, ничего не сказано про лин. независимость);
- Теорема: В решётке корней $L$ есть базис из корней $\mathbf{e}_1,\dots,\mathbf{e}_n\in L$: $\forall\ i\neq j\ \mathbf{e}_i\cdot\mathbf{e}_j\in\{0,-1\}$, $\forall\ i\ \mathbf{e}_i\cdot\mathbf{e}_i=2$.

```python
# решётка корней: sqrt2*гексагональная (см. выше, B_hex_sqrt2, Грам = [[2,1],[1,2]])
# ищем все векторы нормы 2 перебором и через short_vectors
pts = lattice_points(matrix(RDF, B_hex_sqrt2), 5)
roots_bruteforce = [p for p in pts if abs(p.dot_product(p) - 2) < 1e-9]
print("перебором:       корней нормы 2 =", len(roots_bruteforce))

L_hex_sqrt2 = IntegralLattice(gram(B_hex_sqrt2))
roots_short = L_hex_sqrt2.short_vectors(3)[2]   # short_vectors(N) -- нормы 0..N-1
print("short_vectors(3)[2]: корней нормы 2 =", len(roots_short))
```

```python
# среди этих 6 корней перебором ищем пару r, r' с r.r' = -1, образующую базис
pts_exact = lattice_points(B_hex_sqrt2, 3)
roots_exact = [p for p in pts_exact if p.dot_product(p) == 2]
print("корней (точно) =", len(roots_exact))

found = None
for r in roots_exact:
    for rp in roots_exact:
        if r.dot_product(rp) == -1 and matrix([r, rp]).det() != 0:
            found = (r, rp)
            break
    if found:
        break

e1, e2 = found
print("e1 =", e1, " e2 =", e2, " e1.e2 =", e1.dot_product(e2))

# проверка: (e1,e2) -- базис той же решётки <=> матрица перехода от (v1,v2) к (e1,e2)
# целочисленна и унимодулярна (det = +-1)
B = matrix(AA, B_hex_sqrt2)
U = matrix(AA, [e1, e2]) * B.inverse()
print("матрица перехода U:\n", U)
print("U целочисленная:", all(u in ZZ for u in U.list()), " |det U| =", abs(U.det()))
```

## Диаграммы Дынкина и теорема Витта

- Построим граф: вершины — $\mathbf{e}_1,\dots,\mathbf{e}_n$, рёбра между $\mathbf{e}_i$ и $\mathbf{e}_j$ — только для $\mathbf{e}_i\cdot\mathbf{e}_j=-1$;

## Решётки и квадратичные формы

- Решётка $\mapsto$ квадратичная форма: если $\mathbf v=\sum x_i \mathbf v_i$, то $Q(\mathbf x)=|\mathbf v|^2=\mathbf v\cdot\mathbf v$;
- $G=BB^{T}=(\mathbf v_i\cdot\mathbf v_j)$ — матрица Грама (симметричная, полож. опр.), $Q(\mathbf x)=\mathbf x^{T}G\,\mathbf x$;
- Замена базиса $B\mapsto UB$, $U\in GL_n(\mathbb{Z})$ $\Rightarrow$ $G\mapsto UGU^{T}$, т.е. $Q(\mathbf x)\mapsto Q(U^{T}\mathbf x)$ — целочисленная замена переменных;
- Обратно: пол. опр. $Q(\mathbf x)=\mathbf x^{T}G\,\mathbf x$ $\to$ разложение $G=BB^{T}$ (напр., Холецкого) $\mapsto$ решётка с базисом из строк $B$;
- Словарь: решётки с точн. до изометрии $\leftrightarrow$ пол. опр. КФ с точн. до целых замен; $\mathrm{vol}(L)^2=\det G$; целая/чётная $\leftrightarrow$ $G$ целая / диагональ $G$ чётная.

```python
# КФ решётки: Q(x) = |v|^2 для v = sum x_i v_i, G = (v_i . v_j) -- матрица Грама.
# Q(x) = x^T G x = sum_i G_ii x_i^2 + 2 sum_{i<j} G_ij x_i x_j -- координата x_i x_j получает
def gram_to_qf(G):
    n = G.nrows()
    entries = [G[i, j] if i == j else 2*G[i, j] for i in range(n) for j in range(i, n)]
    return QuadraticForm(ZZ, n, entries)

Q_Z2 = gram_to_qf(matrix(ZZ, gram(identity_matrix(ZZ, 2))))
print("Z^2: Q(x) =", Q_Z2.polynomial('x'))
show(Q_Z2.polynomial('x'))

Q_hex = gram_to_qf(matrix(ZZ, gram(B_hex_sqrt2)))
print("sqrt2*гексагональная: Q(x) =", Q_hex.polynomial('x'))
show(Q_hex.polynomial('x'))

Q_fcc = gram_to_qf(matrix(ZZ, gram(B_fcc)))
print("FCC (подрешётка Z^3): Q(x) =", Q_fcc.polynomial('x'))
show(Q_fcc.polynomial('x'))
```

## Тета-функция решётки

- Вопрос: сколько векторов решётки $L$ имеют длину $\sqrt{m}$?
  $$r_L(m) = \left|\{\mathbf{v} \in L : |\mathbf{v}|^2 = m\}\right|.$$
- Например, для $L=\mathbb{Z}^k$
  $$r_{\mathbb{Z}^k}(m) = \left|\{\mathbf{x}=(x_1,\dots,x_k)\in\mathbb{Z}^k : x_1^2+\cdots+x_k^2=m\}\right|,$$
- Т.е. $r_{\mathbb{Z}^k}(m) = r_k(m)$ — число представлений $m$ суммой $k$ квадратов.
- $L\cong L' \Rightarrow r_L(m)=r_{L'}(m)$, т.е. инвариант изометрии;
- Для $L\sim L'$ — не инвариант, но число кратчайших векторов — инвариант ($\min_{m\neq 0} r_L(m)$).
- Идея: последовательность чисел $\mapsto$ производящая функция:
  $$(r_L(0), r_L(1), r_L(2), \dots) \ \mapsto\ \sum_{m=0}^\infty r_L(m)\,x^m.$$
- Тета-функция решётки:
  $$\theta_L(z) = \sum_{m=0}^\infty r_L(m)\,q^{m/2} = \sum_{\mathbf{v}\in L} q^{|\mathbf{v}|^2/2},\quad q = e^{2\pi i z};$$
- Лемма: $\theta_L(z)$ — комплексная функция, определена на $\{z \in \mathbb{C} : \mathrm{Im}(z) > 0\}$;
- По построению: $\theta_L(z+1)=\theta_L(z)$.
- Пример: $L=\mathbb{Z}^2$ ($r_2(m)$ — число представлений суммой двух квадратов)
  $$\theta_{\mathbb{Z}^2}(z) = 1 + 4q^{1/2} + 4q + 4q^2 + 8q^{5/2} + \dots = \left(\sum_{n\in\mathbb{Z}} q^{n^2/2}\right)^{\!2} = \theta_{\mathbb{Z}}(z)^2.$$
- Вообще: $\theta_{L_1\perp L_2}=\theta_{L_1}\cdot\theta_{L_2}$.

```python
# тета-функция Z^2 через QuadraticForm
# theta_series() возвращает ряд по степеням q'=q^{1/2}
Q_Z2 = QuadraticForm(ZZ, 2, [1, 0, 1])
theta_Z2 = Q_Z2.theta_series(20)
show(Q_Z2.polynomial('x'))
show(theta_Z2)
```

```python
# проверка theta_{Z^2} = theta_Z^2 перемножением рядов в PowerSeriesRing
prec = 20
R.<q> = PowerSeriesRing(QQ, default_prec=prec)

Q_Z = QuadraticForm(ZZ, 1, [1])
theta_Z = R(Q_Z.theta_series(prec).polynomial())

show(LatexExpr(r"\theta_{\mathbb{Z}}(z):"))
show((theta_Z).add_bigoh(prec))

show(LatexExpr(r"\theta_{\mathbb{Z}}(z)^2:"))
show((theta_Z^2).add_bigoh(prec))

print(R(theta_Z2.polynomial()) == (theta_Z^2).add_bigoh(prec))
```

## Скрытые симметрии тета-функции

- Пусть $L$ — чётная унимодулярная, т.е.
  $|\mathbf{v}|^2$ чётно $\forall\ \mathbf{v} \in L$, $\mathrm{vol}(L)=1$;
- Оказывается, что $\theta_L(-1/z)=z^k\theta_L(z)$;
- И вообще:
  $$\theta_L\!\left(\frac{az+b}{cz+d}\right) = (cz+d)^{k}\,\theta_L(z),$$
  где $a,b,c,d\in\mathbb{Z}$ — любые такие что $ad-bc=1$, а $k$ зависит от размерности $L$;
- Или $g=\begin{pmatrix} a & b \\ c & d\end{pmatrix} \in SL_2(\mathbb{Z})=\{\text{целочисленные матрицы с}\ \mathrm{det} = 1\}$ — группа.
- Определим оператор $\cdot |_g^k$ действующий на функцию $f(z)$:
  $$(f\,|_g^k)(z) = \frac{1}{(cz+d)^k}\ f\left(\frac{az+b}{cz+d}\right);$$
- $(\theta_L\,|_g^k)(z)=\theta_L(z)$ — инвариантна (симметрична) под действием $\cdot |_g^k$, $g\in SL_2(\mathbb{Z})$.

```python
# Численная проверка модулярной инвариантности theta_L для E_8 под действием  z -> -1/z.
# theta_L(z) вычисляем прямо по коэффициентам r_L(m) (theta_series)
# суммироввние q_num = exp(pi*i*z) сходится быстро при не слишком малой Im(z).

def theta_numeric(qf, z, prec=60):
    coeffs = qf.theta_series(prec).list()
    CF = ComplexField(200)
    qnum = (CF(pi) * CF(I) * CF(z)).exp()
    return sum(CF(c) * qnum^m for m, c in enumerate(coeffs))

z0 = CDF(0.3, 0.8)

# E_8: чётная унимодулярная (используем уже построенную выше матрицу Картана G_E8), k=4
Q_E8 = gram_to_qf(G_E8)
lhs_E8 = theta_numeric(Q_E8, -1/z0)
rhs_E8 = (CDF(z0))^4 * theta_numeric(Q_E8, z0)
show(LatexExpr(r"\theta_{\mathbf{E}_8}(-1/z_0)-z_0^4\theta_{\mathbf{E}_8}(z_0):"))
show(z0)
print("|theta(-1/z) - z^4*theta(z)| =", abs(lhs_E8 - rhs_E8))
```

## Модулярные формы: определение

- Модулярная форма веса $k$ относительно группы $\mathrm{SL}_2(\mathbb{Z})$ — это:
  - Комплексная голоморфная функция $f:\mathbb{H}\to\mathbb{C}$ ($\mathbb{H} = \{z \in \mathbb{C} : \mathrm{Im}(z) > 0\}$);
  - Инвариантная относительно $\cdot |_g^k$, т.е.
    $$f\!\left(\frac{az+b}{cz+d}\right) = (cz+d)^k f(z)\ \text{для всех}\ g=\begin{pmatrix}a&b\\c&d\end{pmatrix}\in SL_2(\mathbb{Z});$$
  - $\begin{pmatrix}1&1\\0&1\end{pmatrix} \in SL_2(\mathbb{Z})$ $\Rightarrow$ $f(z+1)=f(z)$ $\Rightarrow$ ряд Фурье по $q = e^{2\pi i z}$
    $$f(z) = \sum_{m=-\infty}^\infty a(m)\,q^m;$$
  - «Голоморфная в $z=\infty$»: $a(m)=0$ при $m<0$ (т.е. $\sum_{m=0}^\infty \dots$)
- $\begin{pmatrix}0&-1\\1&0\end{pmatrix} \in SL_2(\mathbb{Z})$ $\Rightarrow$ $f(-1/z)=f(z)$;
- Лемма: матричная группа $SL_2(\mathbb{Z})$ порождена $T=\begin{pmatrix}1&1\\0&1\end{pmatrix}$ и $S=\begin{pmatrix}0&-1\\1&0\end{pmatrix}$ (т.е. всякая матрица представляется произведением $S$ и $T$ с повторениями).

## Алгебра модулярных форм

- Сумма двух МФ веса k — снова модулярная форма веса k;
- Теорема: Пространство МФ веса $k$ — конечномерное векторное пространство, размерность $d_k$:
  $$
  d_k = \begin{cases}
  0, & k<0\ \text{или}\ k\ \text{нечётно};\\
  \lfloor k/12\rfloor, & k\equiv2\!\!\!\pmod{12};\\
  \lfloor k/12\rfloor+1, & k\geq0\ \text{чётно},\ k\not\equiv2\!\!\!\pmod{12}.
  \end{cases}
  $$
- Лемма: Каждая МФ веса $k$ выражается как конечная сумма некоторых $E_4(z)^i E_6(z)^j$ с $3i+2j=k$
- Произведение двух МФ (весов $k_1$, $k_2$) — снова модулярная форма (веса $k_1+k_2$);
- Теорема: Множество всех МФ — алгебра над $\mathbb{C}$; изоморфна $\mathbb{C}[E_4,E_6]$ (многочлены от $E_4(z), E_6(z)$).

```python
# сверка размерностей d_k с формулой со слайда (k=1..30)
def dk_formula(k):
    if k < 0 or k % 2 != 0:
        return 0
    if k % 12 == 2:
        return k // 12
    return k // 12 + 1

dims = [(k, ModularForms(SL2Z, k).dimension()) for k in range(1, 30)]
print("(k, dim M_k) из Sage:  ", dims)
print("(k, d_k) по формуле:   ", [(k, dk_formula(k)) for k in range(1, 30)])
print("совпадает:", [d for k, d in dims] == [dk_formula(k) for k in range(1, 30)])
print("dim M_k для нечётных = 0, например, k=1,3,5,7:", [ModularForms(SL2Z, k).dimension() for k in [1, 3, 5, 7]])
```

## Арифметическое приложение

- Рассмотрим диофантово уравнение:
  $$(x_1^2+\dots+x_8^2) \;-\; \left(x_1x_3+x_2x_4+x_3x_4+x_4x_5+x_5x_6+x_6x_7+x_7x_8\right) = m;$$
- $\Leftrightarrow$ $Q(\mathbf{x})=2m,\ \mathbf{x}\in\mathbb{Z}^8$. $Q$ — квадратичная форма некоторой решётки в $\mathbb{R}^8$;
- Это решётка $\mathbf{E}_8$;
- $\mathbf{E}_8$ — чётная и унимодулярная $\Rightarrow$ $\theta_{\mathbf{E}_8}(z)$ — МФ веса $8/2=4$;
- Ряд Эйзенштейна $E_4(z)$ — тоже МФ веса $4$;
- Пространство МФ веса $4$ — одномерно $\Rightarrow$ $\theta_{\mathbf{E}_8(z)} = E_4(z)$;
- Приравнивая коэффициенты:
  $$r_{\mathbf{E}_8}(2m) = 240\,\sigma_3(m) = 240 \sum\nolimits_{d\mid m} d^3,\quad \left(r_{\mathbf{E}_8}(2) = 240\text{ — число корней}\right).$$
- Общий случай (чётные унимодулярные $L\subset\mathbb{R}^n$) — асимптотика:
  $$r_L(m)\ \sim\ c_n\cdot\sigma_{n/2-1}(m/2),\ \text{при}\ m\to\infty.$$

```python
# решётка E_8 и её тета-ряд
E8 = IntegralLattice("E8")
Q_E8_std = QuadraticForm(E8.gram_matrix())
prec = 15
theta_E8 = Q_E8_std.theta_series(prec)
show(theta_E8)

# сравнение коэффициентов r_{E8}(2m) с 240*sigma_3(m) и с q-разложением E_4(z)
for m in range(1, prec):
    lhs = theta_E8.list()[m]
    rhs = 240 * sum(d^3 for d in divisors(m))
    print(f"m={m}: r_E8(2m)={lhs}, 240*sigma_3(m)={rhs}, совпадает: {lhs == rhs}")

E4_check = eisenstein_series_qexp(4, prec, normalization='constant')
print("theta_{E8} == E4 (по первым", prec, "коэффициентам):", theta_E8.list() == E4_check.list())
```

## Конгруэнц-подгруппы и сумма четырёх квадратов

- Выше: «МФ веса $k$» — «МФ веса $k$ для группы $SL_2(\mathbb{Z})$»
- Общий случай: $\theta_L(z)$ — модулярная форма для подгруппы $\Gamma\subset SL_2(\mathbb{Z})$:
  $$(\theta_L\,|_g^k)(z)=\psi_g \theta_L(z), \quad g\in\Gamma\quad (\psi_g\text{ — некоторый множитель})$$
- Пример подгруппы:
  $$\Gamma_0(N) = \left\{\begin{pmatrix}a&b\\c&d\end{pmatrix}\in SL_2(\mathbb{Z}) : c\equiv 0 \!\!\!\pmod N\right\};$$
- Для $L=\mathbb{Z}^4$, $\theta_L(z) = \sum_{m=0}^\infty r_4(m)\,q^m$ — модулярная форма веса 2 для $\Gamma_0(4)$;
- Простой базис таких МФ $\Rightarrow$ $r_4(n) = 8\sum_{d\mid n,\ 4\nmid d} d$;
- (Формула $r_2(n) = 4\big(d_1(n)-d_3(n)\big)$ укладывается в эту же картину при некотором обобщении);
- Общая теория МФ $\Rightarrow$ асимптотики для числа решений уравнения $Q(\mathbf{x})=m$, где $Q$ — целочисленная положительно определённая квадратичная форма от $n$ переменных (т.е. $L$ — целая решётка ранга $n$, не обязательно чётная и унимодулярная).

```python
# порождающие Gamma0(4) и размерность соответствующего пространства МФ веса 2
show(Gamma0(4))
show(Gamma0(4).gens())
M2_G4 = ModularForms(Gamma0(4), 2)
print("dim M_2(Gamma0(4)) =", M2_G4.dimension())
```

```python
# theta_{Z^4}(z) -- МФ веса 2 для Gamma0(4)
# проверка классическая формула Якоби для первых n
# r_4(n) = 8 * sum_{d|n, 4 не делит d} d
prec = 20
Q_Z4 = QuadraticForm(ZZ, 4, [1, 0, 0, 0, 1, 0, 0, 1, 0, 1])
show(Q_Z4.polynomial('x'))
theta_Z4 = Q_Z4.theta_series(prec)
show(theta_Z4)

for n in range(1, prec):
    lhs = theta_Z4.list()[n]
    rhs = 8 * sum(d for d in divisors(n) if d % 4 != 0)
    print(f"n={n}: r_4(n)={lhs}, 8*sum(d|n,4∤d)d={rhs}, совпадает: {lhs == rhs}")
```

## Тета-функция — не полный инвариант

- $L\cong L' \Rightarrow \theta_L=\theta_{L'}$, но обратное не верно;
- Пример: $L=\mathbf{E}_8\perp\mathbf{E}_8$ и $L'=\mathbf{D}_{16}^{+}$ (строится по $\mathbf{D}_{16}$) — две чётные унимодулярные ранга 16, но не изометричны;
- Аналогично выше: вес $\theta$ равен $16/2=8$, пр-во МФ веса $8$ одномерно,
  $$\theta_{\mathbf{E}_8\perp\mathbf{E}_8}(z) = \theta_{\mathbf{D}_{16}^{+}}(z) = E_4(z)^2;$$
- Сколько (различных) классов изометрии чётных унимодулярных решёток ранга $n$?
  - $n=8$: $\#=1$ ($\mathbf{E}_8$);
  - $n=16$: $\#=2$ ($\mathbf{E}_8\perp\mathbf{E}_8, \mathbf{D}_{16}^{+}$);
  - $n=24$: $\#=24$ (решётки Нимайера: у 23 есть корневая подрешётка и одна без корней — решётка Лича);
  - $n=32$: $\# \geqslant 8\cdot 10^7$ (оценка из теории МФ);
  - …

```python
# D_16^+ - максимальная чётная надрешётка решётки корней D_16 
# (эта надрешётка единственна с точностью до изометрии и унимодулярна)
L_D16 = IntegralLattice(['D', 16])
G_D16p = matrix(ZZ, L_D16.maximal_overlattice().gram_matrix())
print("det G(D_16^+) =", G_D16p.det(), " чётная:", all(G_D16p[i, i] % 2 == 0 for i in range(16)))

# G_E8E8 уже построена выше (пример решёток корней ранга 16)
Q_E8E8 = QuadraticForm(G_E8E8)
Q_D16p = QuadraticForm(G_D16p)

prec = 10
theta_E8E8 = Q_E8E8.theta_series(prec)
theta_D16p = Q_D16p.theta_series(prec)
show(theta_E8E8)
show(theta_D16p)
print("theta_{E8+E8} == theta_{D16+}:", theta_E8E8.list() == theta_D16p.list())

E4 = eisenstein_series_qexp(4, prec, normalization='constant')
print("theta_{E8+E8} == E4^2:", theta_E8E8.list() == (E4^2).list())
```

## От решёток к упаковкам шаров, ячейка Вороного, упаковки и покрытия

- $\lambda_1(L)=\min_{\mathbf v\in L\setminus\{0\}}|\mathbf v|$ — длина кратчайшего ненулевого вектора («первый минимум»);
- Упаковка, отвечающая $L$: шары радиуса $\rho(L)=\tfrac12\lambda_1(L)$ с центрами в точках $L$ не пересекаются;
- Контактное число (kissing number): $\tau(L)=\big|\{\mathbf v\in L:|\mathbf v|=\lambda_1(L)\}\big|=r_L(\lambda_1(L)^2)$;
- Ячейка Вороного: $V(L)=\{\mathbf x\in\mathbb{R}^n:|\mathbf x|\le|\mathbf x-\mathbf v|\ \forall\,\mathbf v\in L\}$ — выпуклый многогранник, фундаментальная область (сдвиги $V(L)+\mathbf v$, $\mathbf v\in L$, замощают $\mathbb{R}^n$), $\mathrm{vol}(V(L))=\mathrm{vol}(P(L))=\mathrm{vol}(L)$;
- $\rho(L)$ — радиус наибольшего шара, вписанного в $V(L)$ (радиус упаковки); $\mu(L)$ — радиус наименьшего шара, содержащего $V(L)$ (радиус покрытия); $\rho(L)\leq\mu(L)$;
- Плотность упаковки $\Delta(L)=V_n\,\rho(L)^n/\mathrm{vol}(L)$, плотность покрытия $\Theta(L)=V_n\,\mu(L)^n/\mathrm{vol}(L)$, где $V_n=\pi^{n/2}/\Gamma(n/2+1)$ — объём единичного шара в $\mathbb{R}^n$; $\Delta(L)\leqslant1\leqslant\Theta(L)$.

```python
# ячейка Вороного точки 0 -- как многогранник (Polyhedron), через готовую VoronoiDiagram
from sage.geometry.voronoi_diagram import VoronoiDiagram

def voronoi_cell_at_origin(B, N):
    pts = lattice_points(matrix(RDF, B), N)
    V = VoronoiDiagram([tuple(p) for p in pts])
    for k, poly in V.regions().items():
        if vector(RDF, tuple(k)).norm() < 1e-6:
            return poly
    raise ValueError("ячейка точки 0 не найдена -- увеличьте N")
```

```python
# таблица rho(L), Delta(L) для решёток корней ADE (n=2..8): кратчайшие векторы решётки
# корней -- сами корни (|r|^2=2), поэтому lambda_1=sqrt(2) и rho=1/sqrt(2) одинаковы для всех
def rho_and_delta(cartan_type):
    L, G = root_lattice_from_type(cartan_type)
    n = G.nrows()
    rho = sqrt(2) / 2
    vol = sqrt(G.det())
    Vn = pi**(n/2) / gamma(n/2 + 1)
    Delta = Vn * rho**n / vol
    return rho.n(), Delta.n()

for n in range(2, 9):
    rho, Delta = rho_and_delta(['A', n])
    print(f"A_{n:<2d} rho = {rho:.4f}  Delta = {Delta:.4f}")
for n in range(4, 9):
    rho, Delta = rho_and_delta(['D', n])
    print(f"D_{n:<2d} rho = {rho:.4f}  Delta = {Delta:.4f}")
for n in [6, 7, 8]:
    rho, Delta = rho_and_delta(['E', n])
    print(f"E_{n:<2d} rho = {rho:.4f}  Delta = {Delta:.4f}")
```

```python
# rho(L), Delta(L) для решёток корней ранга 16: две неприводимые (A_16, D_16) и несколько
# приводимых прямых сумм того же ранга. rho=1/sqrt(2) у всех (это всё ещё решётки корней,
# кратчайшие векторы -- корни нормы 2); Delta зависит только от объёма (чем он больше, тем
# решётка "разреженнее" при том же множестве длин корней)
def rho_and_delta_from_gram(G, n):
    rho = sqrt(2) / 2
    vol = sqrt(G.det())
    Vn = pi**(n/2) / gamma(n/2 + 1)
    Delta = Vn * rho**n / vol
    return rho.n(), Delta.n()

G_A16 = matrix(ZZ, CartanMatrix(['A', 16]))
G_D16 = matrix(ZZ, CartanMatrix(['D', 16]))
G_E8 = matrix(ZZ, CartanMatrix(['E', 8]))
G_D8 = matrix(ZZ, CartanMatrix(['D', 8]))
G_A8 = matrix(ZZ, CartanMatrix(['A', 8]))
G_A1 = matrix(ZZ, CartanMatrix(['A', 1]))

lattices_16 = [
    ("A_16 (неприв.)",        G_A16),
    ("D_16 (неприв.)",        G_D16),
    ("E_8 (+) E_8",           block_diagonal_matrix(G_E8, G_E8)),
    ("D_8 (+) D_8",           block_diagonal_matrix(G_D8, G_D8)),
    ("A_8 (+) A_8",           block_diagonal_matrix(G_A8, G_A8)),
    ("E_8 (+) D_8",           block_diagonal_matrix(G_E8, G_D8)),
    ("A_1^{(+)16}",           block_diagonal_matrix(*([G_A1] * 16))),
]

for name, G in lattices_16:
    rho, Delta = rho_and_delta_from_gram(G, 16)
    print(f"{name:16s} vol = {sqrt(G.det()).n():10.4f}  rho = {rho:.4f}  Delta = {Delta:.6f}")
```
